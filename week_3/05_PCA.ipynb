# PCA

Principal Component Analysis replaces the 768 embedding values with a smaller set of new
numbers, chosen so that **as much of the variance as possible is kept**.

No labels and PCA does not know what the images are. This is a
dimensionality reduction method and not a classifier.

How much can we throw away, and does what is left still help us for clustering?

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA

DATA = "../week_2/data"

d = np.load(f"{DATA}/embeddings_stl10_1000.npz")

X = d["E_l12"]
y = d["y"]

class_names = [l.strip() for l in open(f"{DATA}/stl10_binary/class_names.txt") if l.strip()]

print(X.shape)
```

### How much variance is in how many components?

```python
pca = PCA(n_components=50, random_state=0).fit(X)

ratio = pca.explained_variance_ratio_
cumulative = np.cumsum(ratio)

for k in (2, 5, 10, 32, 50):
    print(f"{k:3} components -> {cumulative[k - 1]:.1%} of the variance")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 3.4))

axes[0].bar(range(1, 51), ratio * 100)
axes[0].set_xlabel("component")
axes[0].set_ylabel("% variance explained")
axes[0].set_title("scree plot")

axes[1].plot(range(1, 51), cumulative * 100, marker=".")
axes[1].axhline(90, color="grey", ls="--", lw=1)
axes[1].set_xlabel("components kept")
axes[1].set_ylabel("% cumulative variance")
axes[1].set_title("how much do we keep")

plt.tight_layout()
plt.show()
```

### Project to 2 dimensions

The first two components, drawn as a scatter and coloured by the true class.

The colours are **not** an input to PCA; they are only there so we can see whether the
projection happened to line up with the ten categories.

```python
Z2 = PCA(n_components=2, random_state=0).fit_transform(X)

fig, ax = plt.subplots(figsize=(7, 5.5))

for c in range(1, 11):
    m = y == c
    ax.scatter(Z2[m, 0], Z2[m, 1], s=18, label=class_names[c - 1])

ax.set_xlabel("1st component")
ax.set_ylabel("2nd component")
ax.legend(fontsize=7, ncol=2, loc="best")
plt.tight_layout()
plt.show()
```

### The same in 3 dimensions

```python
Z3 = PCA(n_components=3, random_state=0).fit_transform(X)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

for c in range(1, 11):
    m = y == c
    ax.scatter(Z3[m, 0], Z3[m, 1], Z3[m, 2], s=12, label=class_names[c - 1])

ax.set_xlabel("PC 1")
ax.set_ylabel("PC 2")
ax.set_zlabel("PC 3")
ax.legend(fontsize=7, ncol=2)
plt.tight_layout()
plt.show()
```
