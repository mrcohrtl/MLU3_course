# Synthetic blobs

Unsupervised learning starts with data that has **no labels**.

To practise we make our own data, where we *do* know the right answer, and then we
deliberately hide it. `make_blobs` generates Gaussian clouds: we control how many there
are, how spread out they are, and how different they are from each other.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

RANDOM_STATE = 0

X_clean, y_clean = make_blobs(
    n_samples=300,
    centers=4,
    n_features=2,
    cluster_std=[0.6, 1.2, 0.6, 1.8],   # not all clusters are equally spread out
    center_box=(-10.0, 10.0),
    random_state=RANDOM_STATE,
)

print(f"X_clean : {X_clean.shape}   (n_samples, n_features)")
print(f"y_clean : {y_clean.shape}   the hidden ground truth")
print(f"clusters: {np.unique(y_clean)}")
```

### Features

Two columns of numbers. There is nothing else: no labels, no cluster count, no
information about which point belongs where.

```python
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(X_clean[:, 0], X_clean[:, 1], s=25, c="0.35",
           edgecolor="white", linewidth=0.5)
ax.set_title("The data, seen later by the algorithm")
ax.set_xlabel("feature 1")
ax.set_ylabel("feature 2")
plt.tight_layout()
plt.show()
```

### The ground truth

Four clusters

```python
fig, ax = plt.subplots(figsize=(6, 5))
colors = ["tab:blue", "tab:orange", "tab:green", "tab:red"]

for k in np.unique(y_clean):
    mask = y_clean == k
    ax.scatter(X_clean[mask, 0], X_clean[mask, 1], s=25, c=colors[k],
               edgecolor="white", linewidth=0.5, label=f"cluster {k}")

ax.legend(title="ground truth")
ax.set_title("The same data, with the labels we are hiding")
plt.tight_layout()
plt.show()
```

### Outliers are included

Real measurement data contains points that belong to no cluster at all. We add 15 of
them, far outside the clouds.

We keep **both** versions of the data from here on: the clean one and the one with
outliers. The notebooks afterwards run on each and compare.

```python
rng = np.random.default_rng(RANDOM_STATE)

outliers = rng.uniform(
    low=X_clean.min(axis=0) - 8,
    high=X_clean.max(axis=0) + 8,
    size=(15, 2),
)

X_out = np.vstack([X_clean, outliers])
y_out = np.concatenate([y_clean, np.full(len(outliers), -1)])

print(f"X_out: {X_out.shape}   ({len(X_clean)} clustered + {len(outliers)} outliers)")
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(X_out[:, 0], X_out[:, 1], s=25, c="0.35",
           edgecolor="white", linewidth=0.5)
ax.scatter(outliers[:, 0], outliers[:, 1], s=70, marker="x",
           c="crimson", label="outliers")
ax.legend()
ax.set_title("Same blobs, plus outliers")
plt.tight_layout()
plt.show()
```

### Save

Both datasets are written to `data/blobs.npz` so that every following notebook loads
exactly the same numbers.

```python
from pathlib import Path

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

np.savez(
    data_dir / "blobs.npz",
    X_clean=X_clean, y_clean=y_clean,
    X_out=X_out, y_out=y_out,
)

print("saved:", data_dir / "blobs.npz")
```
